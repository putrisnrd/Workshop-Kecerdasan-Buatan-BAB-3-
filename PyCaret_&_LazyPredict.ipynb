```python
import lazypredict
from lazypredict.Supervised import LazyClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

# 1. Load dataset Breast Cancer
X, y = load_breast_cancer(return_X_y=True)

# 2. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# 3. Fit LazyClassifier
clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)
models, predictions = clf.fit(X_train, X_test, y_train, y_test)

# 4. Tampilkan hasil
print(models)
```

EXAMPLE 3.28C

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(models.index, models['Accuracy'])
```

```python
import lazypredict
from lazypredict.Supervised import LazyClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

X, y = load_breast_cancer(return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)
models, predictions = clf.fit(X_train, X_test, y_train, y_test)

print(models)
```

MODIFIKASI 3.28C

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(models.index, models['Accuracy'])
plt.xticks(rotation=90)
plt.title('Akurasi Model Klasifikasi (Breast Cancer)')
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.tight_layout()
plt.show()
```

MODIFIKASI 3.28D

```python
import lazypredict
from lazypredict.Supervised import LazyRegressor
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split

X, y = load_diabetes(return_X_y=True, as_frame=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=1
)

reg = LazyRegressor(verbose=0, ignore_warnings=True, custom_metric=None)
models_reg, predictions_reg = reg.fit(X_train, X_test, y_train, y_test)

print(models_reg)
```

MODIFIKASI 3.28E

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(models_reg.index, models_reg['R-Squared'], '-s', color='r', linewidth=2)
plt.xticks(rotation=90)
plt.title('R-Squared Model Regresi (Diabetes Dataset)', fontsize=14)
plt.xlabel('Model', fontsize=12)
plt.ylabel('R-Squared', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
